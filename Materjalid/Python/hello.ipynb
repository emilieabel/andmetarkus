## Funktsioonid

```python
name = "Ludwig"

def call_dog(dog_name: str, gender: str, should_get_cookie: bool = None):
    if should_get_cookie == None:
        print("Kuule, see küpsise asi ka, palun vaata, kas veel on küpsist alles.")

    print(f"{dog_name} siia!")
    print("Istu!")
    print(f"Tubli {gender}!")

    if should_get_cookie:
        print("Küpsis!")
```

```python
call_dog("Ludwig", "poiss")
```

```python
call_dog("Ludwig", "poiss", should_get_cookie=False)
```

```python
call_dog(dog_name="Ludwig", gender="poiss")
```

```python
call_dog(gender="poiss", dog_name="Ludwig")
```

```python

def feed_my_dog(dog_name: str, change_water: bool = False):
    print(f"{dog_name}, sinu toit on valmis!")
    if change_water:
        print("Panin uue vee ka sulle.")


feed_my_dog("Ferdinand", change_water=False)
# Ferdinand, sinu toit on valmis!
# Panin uue vee ka sulle.
```
